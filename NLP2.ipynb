```python
from sklearn.preprocessing import MultiLabelBinarizer

sentences = [
    ["I", "love", "Python"],
    ["I", "love", "NLP"],
    ["I", "learn", "Python"]
]

# Create encoder
encoder = MultiLabelBinarizer()

# Convert words into one-hot representation
encoded = encoder.fit_transform(sentences)

print("Vocabulary:")
print(encoder.classes_)

print("\nOne-Hot Encoding:")
print(encoded)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

sentences = [
    "I love Python",
    "I love NLP",
    "I learn Python"
]

vectorizer = CountVectorizer()

# Convert text into numerical representation
X = vectorizer.fit_transform(sentences)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBag of Words:")
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

sentences1 = [
    "I love Python",
    "I love NLP"
]

vectorizer = CountVectorizer(ngram_range=(1, 1))#ngram_range=(minimum, maximum)

X = vectorizer.fit_transform(sentences1)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

sentences2 = [
    "machine learning is powerful",
    "machine learning is useful",
    "deep learning is powerful"
]

vectorizer = CountVectorizer(ngram_range=(2, 2))

X = vectorizer.fit_transform(sentences2)

print("Bigrams:")
print(vectorizer.get_feature_names_out())

print("\nMatrix:")
print(X.toarray())
```
